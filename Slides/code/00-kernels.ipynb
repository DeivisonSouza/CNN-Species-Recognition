```python
import numpy as np
from scipy.signal import convolve2d
from PIL import Image

def aplicar_filtro(imagem, kernel):
    """
    Aplica um kernel (filtro) em uma imagem.
    
    Parâmetros:
    - imagem: PIL.Image ou matriz NumPy da imagem.
    - kernel: matriz NumPy representando o filtro.
    
    Retorna:
    - imagem filtrada como matriz NumPy.
    """
    if isinstance(imagem, Image.Image):
        imagem = np.array(imagem.convert('L'))  # Converter para escala de cinza
    
    # Aplica o filtro usando convolução 2D
    resultado = convolve2d(imagem, kernel, mode='same', boundary='symm')
    
    # Clipa os valores para o intervalo válido
    resultado = np.clip(resultado, 0, 255).astype(np.uint8)
    return resultado

# Filtros comuns usados em CNNs
filtros = {
    'deteccao_bordas_sobel_horizontal': np.array([[ -1, -2, -1],
                                                 [  0,  0,  0],
                                                 [  1,  2,  1]]),
    'deteccao_bordas_sobel_vertical': np.array([[ -1, 0, 1],
                                               [ -2, 0, 2],
                                               [ -1, 0, 1]]),
    'detectar_linhas_vertical': np.array([[ -1, 2, -1],
                                           [ -1, 2, -1],
                                           [ -1, 2, -1]]),
    'detectar_linhas_horizontal': np.array([[ -1, -1, -1],
                                             [  2,  2,  2],
                                             [ -1, -1, -1]]),
    'suavizacao_gaussian_blur': (1/16) * np.array([[1, 2, 1],
                                                    [2, 4, 2],
                                                    [1, 2, 1]]),
    'laplaciano': np.array([[ 0, -1,  0],
                              [-1, 4, -1],
                              [ 0, -1,  0]]),
    'detalhamento': np.array([[ 0, -1,  0],
                                [-1, 5, -1],
                                [ 0, -1,  0]]),
                                'detectar_cantos': np.array([[ -1, -1, -1],
                                 [ -1,  8, -1],
                                 [ -1, -1, -1]]),
    'detectar_curvas': np.array([[ 0, 1, 0],
                                  [ 1, -4, 1],
                                  [ 0, 1, 0]]),
    'detectar_bordas_diagonais': np.array([[ 1,  0, -1],
                                            [ 0,  0,  0],
                                            [-1,  0,  1]]),
    'detectar_bordas_diagonais_2': np.array([[ 0,  1,  0],
                                              [ 1, -4,  1],
                                              [ 0,  1,  0]]),
    'detectar_textura_mais_fina': np.array([[ -1, -1, -1],
                                            [ -1, 8, -1],
                                            [ -1, -1, -1]]),
    'prewitt_horizontal': np.array([[ -1, -1, -1],
                                    [  0,  0,  0],
                                    [  1,  1,  1]]),
    'prewitt_vertical': np.array([[ -1, 0, 1],
                                  [ -1, 0, 1],
                                  [ -1, 0, 1]]),
}
```

```python
# deteccao_bordas_sobel_horizontal
if __name__ == "__main__":
    img = Image.open('image.png')
    filtro_escolhido = 'deteccao_bordas_sobel_horizontal'
    resultado = aplicar_filtro(img, filtros[filtro_escolhido])
    resultado_imagem = Image.fromarray(resultado)
    resultado_imagem.show()
    resultado_imagem.save('sobel_horizontal.png')
```
