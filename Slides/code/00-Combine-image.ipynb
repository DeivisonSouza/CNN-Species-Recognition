```python
#-----------------------------------------------------------------------------#
# Carregar as bibliotecas
#-----------------------------------------------------------------------------#

import os, glob
import math
from PIL import Image, ImageDraw, ImageFont
```

```python
#-----------------------------------------------------------------------------#
# Carregar as imagens, escreve o código e salva e uma lista
#-----------------------------------------------------------------------------#

PATH = './Sample/*.png'
path = glob.glob(PATH, recursive = True)
list = []

for i in path:
    #print(i)
    filename = os.path.splitext(i)[0]
    #print(filename)
    filename_path = os.path.splitext(i)[0]
    code = os.path.split(filename_path)[-1].split('_')
    print(code)
    img = Image.open(i)
    #print(img)
    draw = ImageDraw.Draw(img)
    #print(draw)
    Class = int(code[0])
    I1 = ImageDraw.Draw(img)
    myFont = ImageFont.truetype("C:\Windows\Fonts\\arialbi.ttf", 300)
    I1.text((50, 20), str(Class), font = myFont, fill = 'white')
    #img.save("car2.png")
    list.append(img)
    
```

```python
#-----------------------------------------------------------------------------#
# Função para combinar imagens e salvar
#-----------------------------------------------------------------------------#

def combine(im_list, n_rows, resample=Image.BICUBIC):
    min_height = min(im.height for im in im_list)

    im_list_resize = [
        im.resize((int(im.width * min_height / im.height), min_height), resample=resample) for im in im_list
    ]

    max_width = max(im.width for im in im_list_resize)

    n_cols = math.ceil(len(im_list) / n_rows)
    total_width = (max_width) * n_cols  # compute total width
    total_height = (min_height) * n_rows

    dst = Image.new("RGB", (total_width, total_height))  # calculate the total height

    pos_x = 1
    pos_y = 1
    for i in range(len(im_list_resize)):
        im = im_list_resize[i]
        dst.paste(im, (pos_y, pos_x))

        pos_y += im.width

        if (i + 1) % n_cols == 0:
            # new row
            pos_x += im.height
            pos_y = 0

    print(" total: ", total_width, total_height)
    return dst
```

```python
#-----------------------------------------------------------------------------#
# Combina as imagens e salva
#-----------------------------------------------------------------------------#

combine(im_list=[
                list[0], # 1
                list[1], # 2
                list[2], # 3
                list[3], # 4
                list[4], # 5
                list[5], # 6
                list[6], # 7
                ], n_rows = 4).save('./Sample/Combine/Bark-paleta.jpg')
```
